```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\KC\OneDrive\Desktop\ecommerce-analytics-project\DataSet\Superstore.csv",
    encoding="latin1"
)

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df=df.drop_duplicates()
```

```python
df["Profit_Margin"]=(df["Profit"]/df["Sales"])*100
```

```python
df["Month"] = df["Order Date"].dt.month
```

```python
region_sales=df.groupby("Region")["Sales"].sum()
print(region_sales)
```

```python
category_wise_profit=df.groupby("Category")["Profit"].sum()
print(category_wise_profit)
```

```python
top_customer=df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)
print(top_customer)
```

```python
monthly_sales = (
    df.groupby("Month")["Sales"]
    .sum()
)
print(monthly_sales)
```

```python
#RFM Code
latest_date = df["Order Date"].max()

rfm = df.groupby("Customer ID").agg({
    "Order Date": lambda x: (latest_date - x.max()).days,
    "Order ID": "count",
    "Sales": "sum"
})

rfm.columns = ["Recency", "Frequency", "Monetary"]

print(rfm.head())
```

```python
#FORECASTING
monthly_sales = df.groupby("Month")["Sales"].sum().reset_index()

monthly_sales["Forecast"] = (
    monthly_sales["Sales"]
    .rolling(3)
    .mean()
)
print(monthly_sales)
```
